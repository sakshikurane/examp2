```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns

from sklearn.ensemble import AdaBoostClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from sklearn.preprocessing import LabelEncoder
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("zeyadmohamed26/netflix-customer-churn-and-engagement-analytics")

print("Path to dataset files:", path)
```

```python
df= pd.read_csv(f'{path}/netflix_customer_churn.csv')
df
```

```python
df=df.drop(columns=['customer_id'])
```

```python
df.head(2)
```

```python
obj =LabelEncoder()

df['gender'] =obj.fit_transform(df['gender'])
df['subscription_type'] =obj.fit_transform(df['subscription_type'])
df['region'] =obj.fit_transform(df['region'])
df['device'] =obj.fit_transform(df['device'])
df['payment_method'] =obj.fit_transform(df['payment_method'])
```

```python
df.head()
```

```python
df.columns
```

```python
x=df[['age', 'gender', 'subscription_type', 'watch_hours', 'last_login_days',
       'region', 'device', 'monthly_fee', 'churned', 'payment_method',
       'number_of_profiles', 'avg_watch_time_per_day']]

y=df['favorite_genre']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
model=AdaBoostClassifier()
model.fit(x_train, y_train)
```

```python
predictions = model.predict(x_test)
predictions
```

```python
Score= classification_report(y_test, predictions)
print(Score)
```

```python
import pickle
pickle.dump(model, open('netflix.pkl','wb'))
print('PKL File Created Successfully')
```
